# Dataset exploration

A first look at the two datasets used in the paper. All cleaning lives in `src/nids/data/`; this notebook only calls the loaders, so anything fixed there shows up here.

```python
import matplotlib.pyplot as plt

from nids.data import load_cicids2017, load_nsl_kdd
from nids.data import cicids2017, nsl_kdd
```

## NSL-KDD (Scenario 1)

41 features, split into three simulated data sources — one per heterogeneous detector.

```python
kdd = load_nsl_kdd("train")
print(kdd.shape)
{source: len(cols) for source, cols in nsl_kdd.SOURCES.items()}
```

```python
kdd["category"].value_counts().plot.bar(title="NSL-KDD train: records per category", logy=True)
plt.show()
```

```python
# Top attack labels; teardrop is the paper's worked example
kdd.loc[kdd["is_attack"], "label"].value_counts().head(10)
```

```python
# Each detector only sees its own slice of the features
kdd[nsl_kdd.source_features("connection")].head()
```

## CICIDS2017 (Scenario 2)

The Wednesday (DoS) and Friday-afternoon (DDoS, PortScan) files. Loading takes ~25 s.

```python
cic = load_cicids2017(cicids2017.SCENARIO2_FILES)
print(cic.shape, "|", len(cicids2017.feature_columns(cic)), "features")
cic["label"].value_counts()
```

```python
cic["label"].value_counts().plot.barh(title="CICIDS2017 scenario-2 flows per label", logx=True)
plt.show()
```

```python
# Which detector is trained on which attacks (the paper's Scenario 2 split)
{"detector 1": cicids2017.DETECTOR1_ATTACKS, "detector 2": cicids2017.DETECTOR2_ATTACKS}
```

```python
# Features that never vary carry no information for the classifiers
features = cicids2017.feature_columns(cic)
constant = [c for c in features if cic[c].nunique() == 1]
constant
```
